```python
data = """11-22,95-115,998-1012,1188511880-1188511890,222220-222224,
1698522-1698528,446443-446449,38593856-38593862,565653-565659,
824824821-824824827,2121212118-2121212124
"""

with open("day2.txt", 'r') as f:
   data = f.read()
    
# data = [[y.strip() for y in x.split("-")] for x in data.split(',') if not x == ""]
# print(data)


# Parse data into list of [start, end]
# Removes newlines, splits by comma, filters empties
ranges = [[int(n) for n in x.split('-')] for x in data.replace('\n', '').split(',') if x]

# Find the highest number we need to care about
max_val = max(r[1] for r in ranges)
found = set()

# Generate patterns (seeds) instead of iterating ranges
# e.g., seed "12" -> "1212", "121212"
for seed in range(1, 10**(len(str(max_val))//2 + 1)):
    s_seed = str(seed)
    current_str = s_seed * 2 # Must repeat at least once
    
    while True:
        if len(current_str) > len(str(max_val)): break
        
        val = int(current_str)
        if val > max_val: break

        # Check if generated value falls into ANY target range
        for start, end in ranges:
            if start <= val <= end:
                found.add(val)
                break
        
        current_str += s_seed

print(sum(found))
```
