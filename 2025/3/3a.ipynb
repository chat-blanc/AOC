```python
import numpy as np

with open("input.txt", 'r') as f:
   data = f.read()

# data = [[y.strip() for y in x.split("-")] for x in data.split(',') if not x == ""]
# print(data)

# data = data.split('\n')
# print(data)


data = [x for x in data.split('\n') if not x == ""]
data = [[int(y) for y in x] for x in data ]
data
```

```python
def find_max(row):
    i_1 = np.argmax(row[:-1]) 
    i_2 = np.argmax(row[i_1+1:]) 
    return int(str(row[i_1]) + str(row[i_1+1:][i_2]))

sum([find_max(row) for row in data])
```
