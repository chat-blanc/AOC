```python
import numpy as np

data = """3-5
10-14
16-20
12-18

1
5
8
11
17
32"""

with open("input.txt", 'r') as f:
   data = f.read()

block_1, block_2 = data.strip().split('\n\n')
block_ranges = block_1.splitlines()
block_products = block_2.splitlines()

print(block_ranges)
print(block_products)
```

```python
print(f"Length of the block_products list: {len(block_products)}")
## NO ##
## Create a set out of all the numbers in the block_ranges list
fresh_products = []

for p in block_products:
    print(p)

    for r in block_ranges:
        print(f"Range = {r}")
        x,y = r.strip().split('-')
        print(f"x = {x}, y = {y}")          
        if int(x) <= int(p) <= int(y):
            print("product is fresh")
            fresh_products.append(p)
            break
        else:
            continue

print(len(fresh_products))
```
