```python
import numpy as np

data = """3-5
10-14
16-20
100-104
12-18
17-19
18-21
22-23
49-56

1
5
8
11
17
32"""

    # data = """3-5 = 3
    # 10-14         = 5
    # 16-20
    # 100-104
    # 12-18
    # 17-19
    # 18-21
    # 22-23
    # 49-56


with open("input.txt", 'r') as f:
   data = f.read()

block_1, block_2 = data.strip().split('\n\n')
block_ranges = block_1.splitlines()

range_list = [[int(x) for x in s.split('-')] for s in block_ranges]
range_list.sort()
print(range_list)
```

```python
counter = 1
start_point = range_list[0][0]
print(f"start with {start_point}")

def in_range(range_list,start_point,counter):
    if len(range_list)!=0:
        current_item = range_list.pop(0)
        print(f"current_item is {current_item}")
    else:
        return print(counter)
    ## if the range is smaller than the previous range go to the next one
    if current_item[0] < start_point and current_item[1] <= start_point:
        print("return to function because range is smaller than start_point")
        print(current_item[0], start_point, current_item[1])
        return in_range(range_list,start_point,counter)

    else:
        ## if the current_item is larger or the same as the start point
        ## start the check
        print("start check")
        if current_item[0] > start_point:
            print(f"current_item {current_item} compared to start_point {start_point}")
            start_point = current_item[0]
            counter +=1
            print(f"new start point = {start_point}")
        elif current_item[0] == start_point:
            print("start point stays the same because current_item[0] is start_point")
        elif current_item[1] > start_point: 
            print("start_point stays the same")

    ## if the current item is in the range of the next item
    ## overlap exists
    ## start a new loop with the same start point but the next item
        if len(range_list) == 0:    
            print(f"last item is {current_item}, last start point is {start_point}")
            if current_item[0] < start_point and current_item[1] <= start_point:
                print("return only the counter")
                print(current_item[0], start_point, current_item[1])
                return print(f"finished counter = {counter}")

            else:
                counter += current_item[1] - start_point 
                return print(f"finished counter = {counter}")
        else:
            if range_list[0][0] <= current_item[1] <= range_list[0][1]:
                print("overlap exists")


                ## else if there is no overlap, the range is finished and we start the count

                print(current_item[1], "-", start_point, "=", current_item[1] - start_point)
                counter += current_item[1] - start_point
                print(f"new counter is {counter}")
                start_point = current_item[1]
            else:
                ## after updating the counter we need a new start point
                ## conditions to be met by the start_point:
                ## 1. the start_point is at least current_item[1] + 1

                print(current_item[1], "-", start_point, "=", current_item[1] - start_point)
                counter += current_item[1] - start_point
                print(f"new counter is {counter}")
                start_point = current_item[1]
            print(f"start next range check with {start_point}")
            return in_range(range_list,start_point,counter)





    print(f"new counter = {counter} -- new start_point = {start_point}")
    in_range(range_list,start_point,counter)




print(in_range(range_list,start_point,counter))
```
